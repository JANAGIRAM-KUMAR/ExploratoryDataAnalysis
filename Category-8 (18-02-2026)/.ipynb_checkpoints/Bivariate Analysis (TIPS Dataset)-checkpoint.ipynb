```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression

sns.set(style="whitegrid")
```

```python
df_tips = sns.load_dataset('tips')
df_tips.head()
```

```python
cov_value = df_tips['total_bill'].cov(df_tips['tip'])
print("Covariance:", cov_value)
```

```python
corr_value = df_tips['total_bill'].corr(df_tips['tip'])
print("Correlation:", corr_value)
```

```python
X = df_tips[['total_bill']]
y = df_tips['tip']

model = LinearRegression()
model.fit(X, y)

print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
plt.figure()
plt.scatter(df_tips['total_bill'], df_tips['tip'])
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.title("Scatter Plot - Tips Dataset")
plt.show()
```

```python
plt.figure()
plt.scatter(X, y)
plt.plot(X, model.predict(X))
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.title("Trend Analysis - Tips Dataset")
plt.show()
```
