```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.datasets import load_diabetes

sns.set(style="whitegrid")
```

```python
diabetes = load_diabetes()
df_diabetes = pd.DataFrame(diabetes.data, columns=diabetes.feature_names)

# Add target column (disease progression)
df_diabetes['target'] = diabetes.target

df_diabetes.head()
```

```python
cov_value = df_diabetes['bmi'].cov(df_diabetes['target'])
print("Covariance:", cov_value)
```

```python
corr_value = df_diabetes['bmi'].corr(df_diabetes['target'])
print("Correlation:", corr_value)
```

```python
X = df_diabetes[['bmi']]
y = df_diabetes['target']

model = LinearRegression()
model.fit(X, y)

print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
plt.figure()
plt.scatter(df_diabetes['bmi'], df_diabetes['target'])
plt.xlabel("BMI")
plt.ylabel("Disease Progression")
plt.title("Scatter Plot - Diabetes Dataset")
plt.show()
```

```python
plt.figure()
plt.scatter(X, y)
plt.plot(X, model.predict(X))
plt.xlabel("BMI")
plt.ylabel("Disease Progression")
plt.title("Trend Analysis - Diabetes Dataset")
plt.show()
```
